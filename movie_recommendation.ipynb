```python
import ast
import pandas as pd

# Reload df from file to get the original raw genres column back
df = pd.read_csv('tmdb_5000_merged.csv')
df = df[["title","genres","overview","tagline","vote_average","popularity"]]
df['overview'] = df['overview'].fillna('')

def extract_and_join_genres(genres_str):
    try:
        # Parse stringified list of dictionaries
        genres_list = ast.literal_eval(genres_str)
        # Extract the 'name' key and join them as independent words
        return ' '.join([genre['name'] for genre in genres_list])
    except (ValueError, SyntaxError):
        return ''

# Apply extraction and join
df['genres'] = df['genres'].apply(extract_and_join_genres)
```

```python
df.shape
```

```python
df['tagline'] = df['tagline'].fillna('')
```

```python
df.isnull().sum()
```

```python
df['tags']=df['overview']+df['genres']+df['tagline']
```

```python
df.head()
```

```python
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('wordnet')
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python
import re

def preprocessing(text):
    text = str(text).lower()
    # Remove punctuation using regex
    text = re.sub(r'[\[^\w\s]]', '', text)
    words= text.split()
    words= [word for word in words if word not in stop_words]
    words= [lemmatizer.lemmatize(word) for word in words]
    return ' '.join(words)
```

```python
df['tags'][0]
```

```python
df['tags'] = df['tags'].apply(preprocessing)
```

```python
df['tags'][0]
```

```python
df.head()
```

```python
df =df.reset_index(drop=True)
```

```python
indicies = pd.Series(df.index, index=df['title']).drop_duplicates()
```

```python
indicies
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(max_features=50000,ngram_range=(1,2),stop_words='english')
```

```python
tfidf_matrix = tfidf.fit_transform(df['tags'])
```

```python
tfidf_matrix.shape
```

```python
print(tfidf_matrix.toarray())
```

```python
from sklearn.metrics.pairwise import linear_kernel
```

```python
def recommend(title, n=10):
  if title not in indicies:
    return "Title not found"
  idx= indicies[title]
  sim_core = linear_kernel(tfidf_matrix[idx],tfidf_matrix).flatten()
  similar_idx= sim_core.argsort()[::-1][1:n+1]
  return df['title'].iloc[similar_idx]
```

```python
recommend('The Dark Knight Rises')
```

### Exporting Models & Data as Pickle Files
We will export the processed `df` (so you can load titles easily in your app), the `tfidf_matrix`, and the `tfidf` vectorizer to run recommendations in your production or app environment.

```python
import pickle

# Save the processed DataFrame containing movie titles, IDs, and tags
with open('movie_list.pkl', 'wb') as f:
    pickle.dump(df, f)

# Save the fitted TF-IDF Vectorizer
with open('tfidf_vectorizer.pkl', 'wb') as f:
    pickle.dump(tfidf, f)

# Save the calculated TF-IDF matrix
with open('tfidf_matrix.pkl', 'wb') as f:
    pickle.dump(tfidf_matrix, f)

# Save the movie title-to-index mapping
with open('indices.pkl', 'wb') as f:
    pickle.dump(indicies, f)

```
